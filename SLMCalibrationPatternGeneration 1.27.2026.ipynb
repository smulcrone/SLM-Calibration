```python
import csv
import numpy as np
import pandas as pd
import os
```

```python
def updatePattern(pixelValue):

    #Open CSV
    patternCSV = pd.read_csv('From Santec Pattern Software/.2degree_10bit.csv')

    #Change all values that correspond with the solid region to pixelValue
    patternCSV.iloc[:, 1:961] = pixelValue
    # patternCSV.iloc[:, 1:499] = pixelValue
    # patternCSV.iloc[:, 962:1921] = pixelValue


    maxPixelValue = False

    halfPeriod = 2


    for i in range(962, 1921, halfPeriod):
    # for i in range(500, 1921, halfPeriod):
    # for i in range(1, 961, halfPeriod):


        if(maxPixelValue):
            patternCSV.iloc[:,i:i+halfPeriod] = 1023
            maxPixelValue = False
        
        else:
            patternCSV.iloc[:,i:i+halfPeriod] = 0
            maxPixelValue = True
        

    #Save out CSV with pixelValue in the name so it is distinguishable
    outputFolder = f"CalibrationPattern(p{halfPeriod})"

    if pixelValue < 100:
        outputFilename = f"CalibrationPatternPixelValue(00{pixelValue}).csv"
    elif pixelValue < 1000:
        outputFilename = f"CalibrationPatternPixelValue(0{pixelValue}).csv"
    else:
        outputFilename = f"CalibrationPatternPixelValue({pixelValue}).csv"
    



    try:
        outputPath = os.path.join(outputFolder,outputFilename)
        patternCSV.to_csv(outputPath, index=False)
    except OSError:
        os.mkdir(f"CalibrationPattern(p{halfPeriod})")
        outputPath = os.path.join(outputFolder,outputFilename)
        patternCSV.to_csv(outputPath, index=False)

```

```python
#Create a set off patterns to test
#To 1035 since I want to include max pixel value of 1020. Otherwise for loop only goes to 1005

for i in range(0, 1035, 15):

    updatePattern(i)
```
