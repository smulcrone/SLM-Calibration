```python
import csv
import numpy as np
import pandas as pd
import os
```

```python
def updatePattern(pixelValue, period):

    #Open CSV
    patternCSV = pd.read_csv('PatternTemplate.csv')

    #Change all values that correspond with the solid region to pixelValue
    patternCSV.iloc[:, 1:961] = pixelValue


    maxPixelValue = False

    halfPeriod = int(period/2)

    for i in range(962, 1921, halfPeriod):


        if(maxPixelValue):
            patternCSV.iloc[:,i:i+halfPeriod] = 1024
            maxPixelValue = False
        
        else:
            patternCSV.iloc[:,i:i+halfPeriod] = 0
            maxPixelValue = True
        

#-------------------------------------------------------------------------------------------------
    #TODO - UPDATE WITH FOLDER NAME
    outputFolder = "YOUR NAME HERE"
#-------------------------------------------------------------------------------------------------

    outputFilename = f"CalibrationPatternPixelValue({pixelValue}).csv"
    
    outputPath = os.path.join(outputFolder,outputFilename)

    patternCSV.to_csv(outputPath, index=False)
```

```python
def createTemplate():
    template = np.empty((1201, 1921),  dtype= object)
    templatePandas = pd.DataFrame(template)

    templatePandas.iloc[0, 0] = 'Y/X'

    for i in range(0,1200):
        templatePandas.iloc[i+1,0] = i

    for i in range(0,1920):
        templatePandas.iloc[0,i+1] = i

    
    templatePandas.to_csv("PatternTemplate.csv", index = False, header = False)
    
```
